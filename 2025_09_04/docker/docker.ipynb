````python
# Fix: Remove Windows-specific packages from requirements.txt
# Read current requirements and filter out Windows-only packages
with open("requirements.txt", "r") as f:
    lines = f.readlines()

# List of Windows-specific packages to exclude
windows_packages = [
    "pywin32",
    "pyreadline3", 
    "wmi",
    "winreg",
    "msvcrt"
]

# Filter out Windows-specific packages and the problematic spaCy model
filtered_lines = []
for line in lines:
    package_name = line.split("==")[0].split("@")[0].strip()
    if (not any(pkg in package_name.lower() for pkg in windows_packages) and 
        not line.startswith("it_core_news_sm @")):
        filtered_lines.append(line)

# Write the cross-platform requirements
with open("requirements_cross_platform.txt", "w") as f:
    f.writelines(filtered_lines)

print("✅ Created requirements_cross_platform.txt")
print(f"Original requirements: {len(lines)} packages")
print(f"Cross-platform requirements: {len(filtered_lines)} packages")
print(f"Removed {len(lines) - len(filtered_lines)} Windows-specific/problematic packages")
````

````python
# Create final Dockerfile with Python 3.11 and cross-platform requirements
dockerfile_final = """
FROM python:3.11

WORKDIR /app

# Copy the certificate bundle from your local system
COPY cacert.pem /usr/local/share/ca-certificates/cacert.crt
RUN update-ca-certificates

# Set environment variables to use the certificates
ENV REQUESTS_CA_BUNDLE=/etc/ssl/certs/ca-certificates.crt
ENV SSL_CERT_FILE=/etc/ssl/certs/ca-certificates.crt

COPY requirements_cross_platform.txt requirements.txt

# Install packages with additional SSL configuration
RUN pip install --trusted-host pypi.org --trusted-host pypi.python.org --trusted-host files.pythonhosted.org --trusted-host github.com --no-cache-dir -r requirements.txt

COPY app.py .

CMD ["python", "app.py"]
"""

with open("Dockerfile_final", "w") as f:
    f.write(dockerfile_final)
    
print("✅ Dockerfile_final created with:")
print("  - Python 3.11 (compatible with your packages)")
print("  - Cross-platform requirements (no Windows packages)")
print("  - Certificate support for corporate networks")
print("  - Trusted host flags for additional SSL bypass")
````
